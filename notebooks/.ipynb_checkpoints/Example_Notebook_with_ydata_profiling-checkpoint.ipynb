Example Notebook for ydata-profiling with Generated Data

This notebook demonstrates how to use the ydata-profiling library
to generate a comprehensive report on a Pandas DataFrame.
It starts by creating a simple synthetic dataset.

## 1. Import Libraries

```
#

import pandas as pd
from ydata_profiling import ProfileReport
import numpy as np

```

## 2. Generate Simple Data

```
#

# Let's create a Pandas DataFrame with a few different data types
np.random.seed(42)  # for reproducibility

data = {
    'numerical_feature': np.random.rand(100),
    'categorical_feature': np.random.choice(['A', 'B', 'C', 'A', 'B'], size=100),
    'integer_feature': np.random.randint(1, 10, size=100),
    'boolean_feature': np.random.choice([True, False], size=100),
    'missing_feature': np.random.choice([1, np.nan, 3], size=100),
    'mixed_feature': [f"value_{i}" if i % 3 == 0 else i for i in range(100)]
}

df = pd.DataFrame(data)

print("First few rows of the generated data:")
print(df.head())

```

## 3. Generate the Profile Report

Now, we'll use the `ProfileReport` class from `ydata-profiling`
to generate a detailed report on our DataFrame.

```
#

profile = ProfileReport(df,
                        title="Simple Generated Data Profile",
                        explorative=True,
                        html={'style': {'full_width': True}})

# The `title` argument sets the title of the report.
# `explorative=True` enables more detailed and advanced computations.
# `html={'style': {'full_width': True}}` helps the report to span the entire width
# of the Jupyter Notebook output.

```

## 4. Display the Report in the Notebook

Simply displaying the `profile` object will render the interactive report
directly within your Jupyter Notebook output.

```
#

profile
```
